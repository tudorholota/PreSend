```python
from pathlib import Path
import joblib
```

```python
root_dir = Path.cwd().parent
model_path = root_dir / "models" / "message_classifier.joblib"

#print(model_path.exists())
```

```python
model = joblib.load(model_path)
```

```python
#text = "Esti un idiot si nu ai nici o sansa sa reusesti in viata. Nu meriti sa traiesti." 
text = "desteptule"
```

```python
prediction = model.predict([text])
print(f"Predictions:", prediction)
```

```python
probabilities = model.predict_proba([text])[0]

classes = model.named_steps["classifier"].classes_

results = sorted(zip(classes, probabilities), key = lambda item: item[1], reverse = True)

for label, probability in results:
    print(f"{label:12} {probability * 100:6.2f}%")
```
