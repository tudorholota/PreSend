```python
from pathlib import Path
import joblib
from datasets import load_from_disk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.pipeline import Pipeline
```

```python
root = Path("train.ipynb").resolve().parents[1]
data_dir = root / "data" / "ro_offense"
model_dir = root / "models"
model_path = model_dir / "message_classifier.joblib"
```

```python
dataset = load_from_disk(str(data_dir))

train = dataset["train"]
test = dataset["test"]

X_train = train["text"]
y_train = train["label"]

X_test = test["text"]
y_test = test["label"]
```

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer(lowercase = True, ngram_range = (1, 2), min_df = 2, max_features = 50000, sublinear_tf = True)),
    ("classifier", LogisticRegression(max_iter = 1000, class_weight = "balanced"))
])
```

```python
model.fit(X_train, y_train)

prediction = model.predict(X_test)

accuracy = accuracy_score(y_test, prediction)
print(f"Accuracy: {accuracy:.4f}:")
```

```python
model_dir.mkdir(parents = True, exist_ok = True)
joblib.dump(model, model_path)
```
